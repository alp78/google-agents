```python
import sys

from google.adk.agents import Agent
from google.adk.runners import InMemoryRunner
from google.adk.tools import google_search
from google.genai import types

from IPython.display import display, HTML
from jupyter_server.serverapp import list_running_servers
```

### Agent Definition

```python
root_agent = Agent(
    name="helpful_assistant",
    model="gemini-2.5-flash-lite",
    description="An agent that can answer general questions.",
    instruction="You are a helpful assistant. Use Google Search for current info or if unsure.",
    tools=[google_search],
)
```

### Run Agent

```python
# Create an InMemoryRunner and tell it to use our root_agent
runner = InMemoryRunner(
    app_name="agents",
    agent=root_agent
)
```

```python
# call the .run_debug() method to send our prompt and get an answer
response = await runner.run_debug("What is AGI")
```
